```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
```

```python

df = pd.read_csv(
    "https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv"
)

df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
```

```python
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['diagnosis']),df['diagnosis'],test_size=0.2)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

```python
y_train
```

```python
X_train
```

```python
X_train_tensor=torch.from_numpy(X_train.astype(np.float32))
X_test_tensor=torch.from_numpy(X_test.astype(np.float32))
y_train_tensor=torch.from_numpy(y_train.astype(np.float32))
y_test_tensor=torch.from_numpy(y_test.astype(np.float32))
```

```python
X_train_tensor.shape
```

```python
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.linear=nn.Linear(num_features,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
      out=self.linear(features)
      out=self.sigmoid(out)
      return out
```

```python
#loss function
loss_fuction=nn.BCELoss()
```

```python
learning_rate=0.1
num_epochs=200
```

```python
model=Model(X_train_tensor.shape[1])
#start the loop
for epoch in range(num_epochs):
  #initalize the forward pass
  y_pred=model(X_train_tensor)
  #loss calculate
  loss=loss_fuction(y_pred,y_train_tensor.view
  )
  #backpass
  loss.backward()

  #update parameter
  with torch.no_grad():
    model.linear.weight-=learning_rate*model.linear.weight.grad
    model.linear.bias-=learning_rate*model.linear.bias.grad

  #after updating the parameters, the weight will turn into zero for another loop
  model.linear.weight.grad.zero_()
  model.linear.bias.grad.zero_()

  #printing the each epoch
  print(f"epoch:{epoch+1},loss:{loss.item()}")
```

```python
model=Model(X_train_tensor.shape[1])
#optim using
optimizer=torch.optim.SGD(model.parameters(),lr=learning_rate)

#start the loop
for epoch in range(num_epochs):

  #initalize the forward pass
  y_pred=model(X_train_tensor)

  #loss calculate
  loss=loss_fuction(y_pred,y_train_tensor.view(-1,1))

  #clearing gradients
  optimizer.zero_grad()

  #backpass
  loss.backward()

  #update the optimizer using optim
  optimizer.step()

  #printing the each epoch
  print(f"epoch:{epoch+1},loss:{loss.item()}")
```

```python
model.linear.bias
```

```python
#model evaluation
with torch.no_grad():
  y_pred=model.forward(X_test_tensor)
  y_pred=(y_pred>0.5
          ).float()
  accuracy=(y_pred==y_test_tensor).float().mean()
  print(f"accuracy:{accuracy.item()}")
```
