```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import torch
```

```python

df = pd.read_csv(
    "https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv"
)

df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
```

```python
df
```

```python
X_train,X_test,y_test,y_train=train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
y_test
```

LABEL ENCODING THE CLASSES

```python
encode=LabelEncoder()
y_train=encode.fit_transform(y_train)
y_test=encode.transform(y_test)
```

COVERSIOON NUMPY ARRAYS TO TENSORS


```python
X_test_tensor=torch.from_numpy(X_test)
X_train_tensor=torch.from_numpy(X_train)
y_test_tensor=torch.from_numpy(y_test)
y_train_tensor=torch.from_numpy(y_train)
```

```python
X_train_tensor.shape
```

MAKING THE NEURAL NETWORK


```python
class my_simple_nn:
  def __init__(self,X):
    self.weights=torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)

    #forward pass
  def forward(self,X):
    #matrix multiplication
    z=torch.matmul(X,self.weights)+self.bias
    y_pred=torch.sigmoid(z)
    return y_pred
#loss calculation
  def loss(self,y_pred,y):
    epsilon=1e-7
    y_pred=torch.clamp(y_pred,epsilon,1-epsilon)
    loss=-(y_train_tensor*torch.log(y_pred)+(1-y_train_tensor)*torch.log(1-y_pred)).mean()
    return loss







```

```python
learning_rate=0.1
epoch=25
```

pipeline


```python

#trainingn pipline
model=my_simple_nn(X_train_tensor)
#dfine loops
for epoch in range(epoch):
  #forward pass
  y_pred=model.forward(X_train_tensor)
  #print(y_pred)

  #calculate loss
  loss=model.loss(y_pred,y_train_tensor)
  #print(f"epoch:{epoch+1},loss:{loss.item()}")

  #backward pass
  loss.backward()


  #parametres update
  with torch.no_grad():#for not tracking the weights
    model.weights-=learning_rate*model.weights.grad
    model.bias-=learning_rate*model.bias.grad

    #zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()
  print(f"epoch:{epoch+1},loss:{loss.item()}")
```

```python
model.bias
```

```python
#model evaluation
with torch.no_grad():
  y_pred=model.forward(X_test_tensor)
  y_pred=(y_pred>0.4
          ).float()
  accuracy=(y_pred==y_test_tensor).float().mean()
  print(f"accuracy:{accuracy.item()}")
```
